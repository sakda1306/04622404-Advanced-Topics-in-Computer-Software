```python
import json
import os
from pathlib import Path
from urllib.request import Request, urlopen
import pandas as pd

os.makedirs("./outputs", exist_ok=True)
```

```python
apis = {
    "aidevboard_ai": "https://aidevboard.com/api/v1/jobs",
    "jobicy_software_en": "https://jobicy.com/api/v2/remote-jobs?industry=engineering",
    "jobicy_devops_infra": "https://jobicy.com/api/v2/remote-jobs?industry=admin",
    "jobicy_cyber_security": "https://jobicy.com/api/v2/remote-jobs?industry=cybersecurity",
}
```

```python
headers = {"Accept": "application/json", "User-Agent": "Mozilla/5.0"}

for k, v in apis.items():
    print(f"{k}: {v}")
    with urlopen(Request(v, headers=headers)) as response:
        jobs_payload = json.load(response)

    job_rows = jobs_payload["jobs"] if isinstance(
        jobs_payload, dict) and "jobs" in jobs_payload else jobs_payload
    Path(f"./outputs/extracted_text_{k}.json").write_text(
        json.dumps(job_rows, indent=2, ensure_ascii=False),
        encoding="utf-8",
    )

    df_jobs = pd.DataFrame(job_rows)
    print(f"{len(df_jobs)} jobs founded in {k}.")
    df_jobs
```
